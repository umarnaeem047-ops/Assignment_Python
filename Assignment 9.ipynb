```python
open_file = open("student data.csv",mode = 'r')
open_file
```

```python
reader = csv.reader(open_file)
reader
```

```python
data = list(reader)
data
```

```python
header = data[0]
header
```

```python
student_data = data[1:]
student_data
```

```python
for percen in student_data:
    print(type(percen[5]))
```

```python
for gra in student_data:
    
    percentage = float(gra[5])
    if percentage >= 80:
                grade = "A"
    elif percentage >= 70:
                grade = "B"
    elif percentage >= 60:
                grade = "C"
    elif percentage >= 50:
                grade = "D"
    else:
                grade = "F"
    gra.append(grade)
```
